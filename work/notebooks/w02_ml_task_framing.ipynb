## 1. My lane as an ML task (type)


## My lane as an ML task

My lane is **Refresh / Content Opportunity Scoring**, which I frame as a **ranking** task.

The objective is to rank existing content pages from highest to lowest refresh priority. Instead of making a simple yes/no decision, the model helps content teams decide which pages should be reviewed first based on their potential opportunity for improvement.

This ranking supports editorial decision-making by focusing attention on the pages with the greatest expected impact.

## 2. Target or proxy



## Target or proxy

There is no direct label that says whether a page should be refreshed, so I would use a **refresh opportunity score** as a proxy target.

The score would be estimated using observed signals such as declining impressions, declining sessions, CTR, average search position, content age, and trend direction. These observed outcomes provide evidence about which pages are more likely to benefit from a refresh.

## 5. Why ML beats a fixed rule here


## Why ML beats a fixed rule

A fixed rule such as "refresh every article older than one year" ignores many important signals.

Some older pages still perform well, while some newer pages lose visibility quickly. ML can learn patterns across many features at the same time, including impressions, CTR, average position, engagement, search volume, competition, and trend direction.

This produces a more useful ranking than a single hand-written rule and supports better editorial decisions.
